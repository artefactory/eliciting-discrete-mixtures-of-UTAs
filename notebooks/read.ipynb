```python
base_dir = "/data/workspace/vincent/elicit_2"
instance_dir = "run_x0"
```

```python
instances = os.listdir(base_dir)
data_sizes = [32, 64, 74, 128, 256, 512, 2048, 4096]
models = ["indiff_coupled", "indiff_singled", "pref_coupled", "pref_singled"]
  

plt.figure(figsize=(12, 10))
for imodel, model in enumerate(models):
    plt.subplot(2, 2, imodel + 1)
    for instance_dir in instances:
        all_res = []
        all_sizes = []
        for ndata in data_sizes:
            try:
                with open(os.path.join(base_dir, instance_dir, f"ndata_{ndata}", model, "fit_params.json"), "r") as res_file:
                    res = json.load(res_file)
                all_res.append(res["optimization_objective"])
                all_sizes.append(ndata)
            except FileNotFoundError:
                pass
        plt.plot(all_sizes, all_res, marker="o", label=instance_dir)
    plt.title(model)
    plt.xlabel("Number of data points")
    plt.ylabel("Optimization objective")
    plt.legend()
```

```python
instances = ["run_run_l0", "run_run_l1", "run_run_l2", "run_run_l3"]
instances = os.listdir(base_dir)
data_sizes = [32, 64, 74, 128, 256, 512, 2048, 4096]
models = ["indiff_coupled", "indiff_singled", "pref_coupled", "pref_singled"]
  

plt.figure(figsize=(6, 4))
for imodel, model in enumerate(models):
    all_means = []
    all_q1 = []
    all_q3 = []
    all_sizes = []
    for ndata in data_sizes:
        curr_values = []
        for instance_dir in instances:
            try:
                with open(os.path.join(base_dir, instance_dir, f"ndata_{ndata}", model, "fit_params.json"), "r") as res_file:
                    res = json.load(res_file)
                curr_values.append(res["optimization_objective"])
            except FileNotFoundError:
                pass
        if len(curr_values) > 0:
            all_means.append(np.mean(curr_values))
            all_q1.append(np.percentile(curr_values, 25))
            all_q3.append(np.percentile(curr_values, 75))
            all_sizes.append(ndata)
    plt.plot(all_sizes, np.array(all_means) / 20, marker="o", label=model)
    plt.fill_between(all_sizes, np.array(all_q1) / 20, np.array(all_q3) / 20, alpha=.5)
plt.title("Average Distance over instances")
plt.xlabel("Number of data points")
plt.ylabel("Optimization objective")
plt.legend()
```
